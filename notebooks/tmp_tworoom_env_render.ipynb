# Temporary TwoRoom Env Render Test

Minimal probe to verify that the true `swm/TwoRoom-v1` Gymnasium environment can be created, reset, and rendered.

```python
import gymnasium as gym
import matplotlib.pyplot as plt
import torch
import stable_worldmodel.envs  # registers swm/* env ids

env = gym.make("swm/TwoRoom-v1", render_mode="rgb_array", disable_env_checker=True)
obs, info = env.reset(seed=0)
obs_np = obs.detach().cpu().numpy() if torch.is_tensor(obs) else obs
frame = env.render()

print("obs type:", type(obs))
print("obs_np type:", type(obs_np))
print("obs_np shape:", getattr(obs_np, "shape", None))
print("action space:", env.action_space)
print("frame shape:", frame.shape, frame.dtype, frame.min(), frame.max())

plt.figure(figsize=(4, 4))
plt.imshow(frame)
plt.axis("off");
```
